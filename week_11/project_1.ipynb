```python
import pandas as pd
from abc import ABC, abstractmethod
```

```python
data = pd.read_csv("SIS.csv")
data
```

```python
class StudentClassifier:
    
    def __init__(self, filename, data):
        self.filename = filename
        self.data = data

    # Method to classify students
    def classify_students(self):

        pirates = []
        yankees = []
        bulls = []

        for index, row in self.data.iterrows():

            age = row["Age"]

            
            if age > 14 and age < 18:
                pirates.append(row)

            
            elif age > 18 and age < 22:
                yankees.append(row)

            
            elif age > 22 and age < 25:
                bulls.append(row)

            # invalid ages
            elif age > 25:
                error = Encapsulation(age)
                print(error.get_error())

        # Convert lists into DataFrames
        pirates_df = pd.DataFrame(pirates)
        yankees_df = pd.DataFrame(yankees)
        bulls_df = pd.DataFrame(bulls)

        # Save to CSV files
        pirates_df.to_csv("The_Pirates.csv", index=False)
        yankees_df.to_csv("The_Yankees.csv", index=False)
        bulls_df.to_csv("The_Bulls.csv", index=False)

        # Display the files
        print("\n=== THE PIRATES ===")
        print(pirates_df)

        print("\n=== THE YANKEES ===")
        print(yankees_df)

        print("\n=== THE BULLS ===")
        print(bulls_df)


# Encapsulation Class
class Encapsulation:

    def __init__(self, age):
        self.__age = age   # private attribute

    # Getter method
    def get_error(self):
        return f"Error: Student age {self.__age} is above the accepted limit."
```

```python
# Main Program
if __name__ == '__main__':
    student_system = StudentClassifier("SIS.csv", data)
    student_system.classify_students()
```

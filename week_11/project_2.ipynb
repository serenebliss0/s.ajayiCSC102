```python
from abc import ABC, abstractmethod
```

```python
class coup_de_escriva(ABC):

    @abstractmethod
    def Fan_page(self):
        pass

class FC_Cirok(coup_de_escriva):
    def Fan_page(self):
        print("\nWelcome to FC Cirok")
        #return super().Fan_page()
    
class Madiba_FC(coup_de_escriva):
    print("\nWelcome to Madiba FC")
    def Fan_page(self):
        return super().Fan_page()
    
class Blue_Jay_FC(coup_de_escriva):
    def Fan_page(self):
        print("\nWelcome to Blue Jay FC")
        return super().Fan_page()
    
class TSG_Walker(coup_de_escriva):
    def Fan_page(self):
        print("\nWelcome to TSG Walker")
        return super().Fan_page()
```

```python
print("1. FC Cirok\n2. Madiba FC\n3. Blue Jay FC\n4. TSG Walker");

while True:
    try:
        user_club = int(input('What club do you support?\n'));
    except Exception as e:
        print(f"Oops! {e}")

    if user_club == 1:
        club = FC_Cirok()
        club.Fan_page()
        break
    elif user_club == 2:
        club = Madiba_FC()
        club.Fan_page()
        break
    elif user_club == 3:
        club = Blue_Jay_FC();
        club.Fan_page()
        break
    elif user_club == 4:
        club = TSG_Walker();
        club.Fan_page()
        break
    else:
        print("Choose a club from the list and try again")
```

```python

# Abstract Class
class Product(ABC):                    
    
    # Normal Method
    def item_list(self, rate):
        print("amount submitted : ",rate)
    
    # Abstract Method
    @abstractmethod
    def product(self,rate):                      
        pass

# Child class
class Item(Product):
    def product(self, rate):
        print("Product rate:", rate)

r = Item()
r.product(3)
r.item_list(2)
```

```python
class geometric(ABC):
    @abstractmethod
    def volume(self):
        #abstract method
        pass
    
class Rect(geometric):
    length = 4
    width = 6
    height= 6
    def volume(self):
        return self.length * self.width *self.height
    
class Sphere(geometric):
    radius = 8
    def volume(self):
        return 1.3 * 3.14 * self.radius * self.radius *self.radius
    
class Cube(geometric):
    Edge = 5
    def volume(self):
        return self.Edge * self.Edge *self.Edge
    
class triangle_3D:
    length = 5
    width = 4
    def volume(self):
        return 0.5 * self.length * self.width
    
rr = Rect()
ss = Sphere()
cc = Cube()
tt = triangle_3D()
print("Volume of a rectangle:", rr.volume())
print("Volume of a circle:", ss.volume())
print("Volume of a square:", cc.volume())
print("Volume of a triangle:", tt.volume())
```

```python
class Invoice(ABC):
    
    def final_bill(self, pay):
        print('Purchase of the product: ', pay)
        
    @abstractmethod
    def invoice(self, pay):
        pass
    
class Paycheque(Invoice):
    def final_bill(self, pay):
        return super().final_bill(pay)
    def invoice(self, pay):
        print('Paycheck of: ', pay)
        
class CardPayment(Invoice):
    def final_bill(self, pay):
        return super().final_bill(pay)
    def invoice(self, pay):
        print('Pay through card of: ', pay)
        
        
class payment():
    def invoice(self, pay):
        print('Pay through card of: ', pay)
    
        
aa = Paycheque()
aa.invoice(6500)
aa.final_bill(6500)
print(isinstance(aa, payment))

aa = CardPayment()
aa.invoice(2600)
aa.final_bill(2600)
print(isinstance(aa, Invoice))
```

```python
class Animal(ABC):

    @abstractmethod
    def move(self):
        pass

class Human(Animal):
    
    def move(self):
        print("I can walk and run")

class Snake(Animal):
    
    def move(self):
        print("I can crawl")

class Dog(Animal):

    def move(self):
        print("I can walk and run")

class Lion(Animal):
    
    def move(self):
        print("I can run")

# Object Instantiation
human = Human()
human.move()

snake = Snake()
snake.move()

dog = Dog()
dog.move()

lion = Lion()
lion.move()
```

```python
import abc
from abc import ABC, abstractmethod

class Zinc(ABC):
    @abstractmethod
    def rk(self):
        print("Class Zinc is an Abstract Base Class")

class Tin(Zinc):
    def rk(self):
        super().rk()
        print("Class Tin is a subclass ")

# Object instantiation
r = Tin()
r.rk()
```

```python
class Bank(ABC):
    def branch(self, Naira):
        print("Fees submitted : ",Naira)

    @abstractmethod
    def Bank(self, Naira):
        pass
    
class private(Bank):
    def branch(self, Naira):
        return super().branch(Naira)
    def Bank(self, Naira):
        print("Total Naira Value here: ",Naira)
        
class public(Bank):
    def branch(self, Naira):
        return super().branch(Naira)
    def Bank(self, Naira):
        print("Total Naira Value here:",Naira)

p = private()
p.Bank(5000)

u = public()
u.Bank(2000)

a = public()
a.branch(3500)
```

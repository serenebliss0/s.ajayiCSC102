```python
from abc import ABC, abstractmethod
import tkinter as tk
from tkinter import messagebox
```

```python
class External_Vendors(ABC):
    
    @abstractmethod
    def menu(self):
        pass
```

```python
class cooperative_cafeteria(External_Vendors):
    def menu(self):
        return """
    Cooperative Cafeteria
    Jollof Rice and Stew - N200
    White Rice and Stew - N200
    Fried Rice - N200
    Salad - N100
    Plantain - N100"""     
```

```python
class faith_cafeteria(External_Vendors):
    def menu(self):
        return """
    Faith Hostel Cafeteria
    Fried Rice - N400
    White Rice and Stew - N400
    Jollof Rice - N400
    Beans - N200
    Chicken - N1000"""     
```

```python
class student_center_cafeteria(External_Vendors):
    def menu(self):
        return """
    Student Center Cafeteria
    Chicken Fried Rice - N800
    Pomo Sauce - N300
    Spaghetti Jollof - N500
    Amala/Ewedu - N500
    Semo with Eforiro Soup - N500"""     
```

```python
def show_menu():
    
    vendor = entry.get().lower()

    if vendor == "cooperative":
        food = cooperative_cafeteria()

    elif vendor == "faith hostel":
        food = faith_cafeteria()

    elif vendor == "student center":
        food = student_center_cafeteria()

    else:
        messagebox.showerror("Error", "Vendor not found!")
        return

    output_label.config(text=food.menu())
```

```python
# GUI Window
root = tk.Tk()
root.title("PAU External Vendors")
root.geometry("500x400")
root.configure(bg="pink")

# Title
title = tk.Label(root, text="PAU Food Vendor System",
                 font=("Arial", 16, "bold"))
title.pack(pady=10)


# Input
label = tk.Label(root, text="Enter Vendor Name:", bg="pink")
label.pack()

entry = tk.Entry(root, width=30)
entry.pack(pady=5)


# Button
button = tk.Button(root, text="Display Menu",
                   command=show_menu)
button.pack(pady=10)


# Output
output_label = tk.Label(root, text="",
                        justify="left",
                        font=("Courier", 11),
                        bg="pink")
output_label.pack(pady=20)


# Run GUI
root.mainloop()
```
